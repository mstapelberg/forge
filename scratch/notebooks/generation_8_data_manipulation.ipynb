This notebook shows how to add new data from many different vasp jobs to your database

```python
from forge.core.database import DatabaseManager
from forge.workflows.vasp_to_db import process_vasp_jobs

# --- Setup ---
# 1. Initialize your DatabaseManager
db_manager = DatabaseManager()

```

```python
new_structures =  db_manager.find_structures_by_metadata({'generation': 8})
print(f"Found {len(new_structures)} new structures")

# Get the first structure
structure = db_manager.get_structure(new_structures[0])
structure_metadata = db_manager.get_structure_metadata(new_structures[0])
print(structure)
print(structure_metadata)
```

```python
db_manager.remove_calculations_batch(new_structures)
db_manager.remove_structures_batch(new_structures)
```

```python
# 2. Define the directories for your VASP jobs
modified_structures_dir = '../data/generation_8_2025-06-17/gen_8_no_shake_rmse_all_vasp_jobs_part1_fin/'
new_elastic_structures_dir = '../data/generation_8_2025-06-17/elastic_jobs_total_finished/'

# --- Execution ---
# Process the modified structures first
# Here, 'config_type' should be present in each metadata.json
print("Processing modified structures...")
process_vasp_jobs(
    db_manager=db_manager,
    base_dir=modified_structures_dir,
    generation_tag=8,
    calculation_type='static' # Assuming these are static calculations
)

# took 2m 57.2s 
```

```python
import json
from pathlib import Path
from tqdm import tqdm

short_range_data = Path('../data/generation_8_2025-06-17/selected_short_range_dimers_data')

print(f"Scanning for OUTCAR files in: {short_range_data}")
outcar_paths = list(short_range_data.rglob('OUTCAR'))
job_dirs = [p.parent for p in outcar_paths]

print(f"Found {len(job_dirs)} directories with OUTCARs. Creating/updating metadata.json...")

for job_dir in tqdm(job_dirs, desc="Processing metadata files"):
    metadata_path = job_dir / "metadata.json"
    
    metadata_content = {
        'config_type': 'short_range_dimer',
        'source_path': str(job_dir.resolve()) # Use resolve() to get absolute path
    }

    try:
        with open(metadata_path, 'w') as f:
            json.dump(metadata_content, f, indent=4)
        # print(f"Created/updated metadata.json in: {job_dir}") # Too verbose for tqdm
    except Exception as e:
        print(f"\n[ERROR] Failed to write metadata.json for {job_dir}: {e}")

print("Finished creating/updating metadata.json files.")

```

```python
from pathlib import Path
from forge.workflows.vasp_to_db import process_vasp_jobs
from forge.core.database import DatabaseManager

db_manager = DatabaseManager()

short_range_data = Path('../data/generation_8_2025-06-17/selected_short_range_dimers_data')
process_vasp_jobs(
    db_manager=db_manager,
    base_dir=short_range_data,
    generation_tag=8,
    calculation_type='static',
    default_config_type='short_range_dimer',
    skip_duplicates=False
)
```

```python
new_structures =  db_manager.find_structures_by_metadata({'generation': 8})
print(f"Found {len(new_structures)} new structures")

# Get the first structure
structure = db_manager.get_structure(new_structures[0])
structure_metadata = db_manager.get_structure_metadata(new_structures[0])
print(structure)
print(structure_metadata)
```
